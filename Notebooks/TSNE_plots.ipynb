```python
# DCA (all) vs non DCA (all)
data_a, classes_a, filepaths_a = gtp.get_all_data(
    data_filepath="/home/sam/Repositories/mmu/balanced_dca_split/data/",
    kernel=(45, 45),
    glob_pattern="/**/clean/**"
)

data_b, classes_b, filepaths_b = gtp.get_all_data(
    data_filepath="/home/sam/Repositories/mmu/balanced_dca_split/data/",
    kernel=(45, 45),
    glob_pattern="/**/dca_binary/**"
)

data_c, classes_c, filepaths_c = gtp.get_all_data(
    data_filepath="/home/sam/Repositories/mmu/balanced_dca_split/data/",
    kernel=(45, 45),
    glob_pattern="/**/dca_realistic/**"
)


pca_features_a = gtp.get_pca_features(data=data_a, n_components=300)
tsne_embeddings_a = gtp.get_tsne_embeddings(pca_features_a, 2, 350, 0.2, 2)
tsne_embeddings_a = gtp.normalize_2d_embeddings(tsne_embeddings_a)

pca_features_b = gtp.get_pca_features(data=data_b, n_components=300)
tsne_embeddings_b = gtp.get_tsne_embeddings(pca_features_b, 2, 350, 0.2, 2)
tsne_embeddings_b = gtp.normalize_2d_embeddings(tsne_embeddings_b)

pca_features_c = gtp.get_pca_features(data=data_c, n_components=300)
tsne_embeddings_c = gtp.get_tsne_embeddings(pca_features_c, 2, 350, 0.2, 2)
tsne_embeddings_c = gtp.normalize_2d_embeddings(tsne_embeddings_c)

gtp.tsne_duo_plot_2(
    8000, 
    6000, 
    50, 
    [filepaths_a, filepaths_b, filepaths_c], 
    [classes_a, classes_b, classes_c], 
    ["clean (mel & oth)", "dca_binary (mel & oth)", "dca_realistic (mel & oth)"], 
    [tsne_embeddings_a, tsne_embeddings_b, tsne_embeddings_c], 
    (20, 35), 
    "../tsne_plots/clean_vs_dca_all.png"
)
```
